# What is python?

Python is a high-level, easy-to-read programming language known for its simplicity and versatility. It's widely used for web development, data analysis, machine learning, and more. Python's clean syntax, extensive standard library, and active community make it a popular choice for both beginners and experts.

Python is cross-platform, available for Windows, macOS, and Linux. You can choose the version you want to install, but it's recommended to use Python 3.8 or later for the latest features and improvements. At the time of this writing, the latest version is 3.12 (released on 2 October 2023), so generally, you don't want to use anything older than 4 major versions and want to wait a bit to use the latest since people who maintain packages need some time to adjust to updates.

# Python installation and virtual environments (Anaconda/Miniconda)
While you can certainly have a system-wide Python installation, it's important to note that different projects often require different Python versions and dependencies. To manage these variations efficiently, we will be using Miniconda or Anaconda. These tools allow us to set up isolated virtual environments with specific Python versions, ensuring a seamless and organized development experience for each project.


## Linux Installation

1. Download the installer:
    - <a href="https://docs.conda.io/en/latest/miniconda.html#linux-installers">Miniconda installer for Linux</a>
    - <a href="https://www.anaconda.com/download/">Anaconda installer for Linux</a>
2. In your terminal window, run:
    - Miniconda: `bash Miniconda3-latest-Linux-x86_64.sh`
    - Anaconda: `bash Anaconda-latest-Linux-x86_64.sh`
3. Follow the prompts on the installer screens.
    - If you are unsure about any setting, accept the defaults. You can change them later.
4. To make the changes take effect, close and then re-open your terminal window.
5. Test your installation in terminal window by running the command: `conda list`.
    - A list of installed packages appears if it has been installed correctly.

## Windows Installation

## Creating Virtual Environment

These instructions guide you through creating, activating, and installing packages in a conda environment, ensuring a clean and organized Python development setup.




### Creating Virtual Environment from scratch

-- on Windows you might need to do:
`conda config --set ssl_verify false`

1. Open Terminal (on Linux) or Anaconda Prompt (on Windows).
<br><br>
2. Creatin Environment
   - Type one of the following commands to create a new conda environment:
    <br> `conda create -n <name>`
    <br> `conda create --name <name>`
        - Replace `<name>` with your custom environment name.
        - This will create an empty conda environment with the specified `<name>`.
    - If you want to use a specific version of Python, add the `python` keyword followed by the desired version:
     <br> `conda create --name <name> python=3.9`
    - After specifying the environment name, you can add any number of packages and their versions, separated by spaces:
     <br> `conda create --name <name> <package> <package>...`
     <br> `conda create --name first_env python=3.9 pandas fastapi=0.68`
<br><br>
3. Activating Environment
    - Before you can use the newly created environment you first need to activate it, use the following command:
     <br> `conda activate <name>`
<br><br>
4. Install Additional Packages
    - Once the environment is activated, you can install additional packages using:
     <br> `conda install <package> <package>...`
     <br> `conda install numpy matplotlib`
    - You can specify the channel when installing packages using conda, and one commonly used channel is Conda Forge (community-led, faster upadtes of packages).
     <br> `conda install -c conda-forge <package> <package>...`
     <br> `conda install -c conda-forge opencv`
       - usually we dont try to mix packages form different channels, so its good to stick only installing packages from one channel if possible
    - You can also use `pip` to install packages that are not available on conda (note that sometimes it can break dependencies)
     <br> `pip install <package> <package>...`
     <br> `pip install numpy matplotlib`
<br><br>
5. Install Packages from files
    - Once the environment is activated, you can install additional packages from requrements files:
     <br> `conda install --file requirements.txt `
     <br> `pip install -r requirements.txt `

### Creating Virtual Environment from YAML file

Use these commands to create and activate the conda environment based on the specifications in the Yaml file:
<br> `conda env create --file environment.yml`
<br> `conda activate first_python_env`

example of yaml file:
```yaml
name: first_python_env
channels:
  - conda-forge
dependencies:
  - python=3.8
  - numpy=1.9.*
  - pandas
```

## Python 3.8
all changes: https://docs.python.org/3/whatsnew/3.8.html

### 1. Positional Only Parameters
Now you can, by using the slash character(/). Parameters defined before the / become positional-only parameters:

````python
def add_a_and_b(a, b, /):
    return a + b

add_a_and_b(3, 5)
````

````python
try:
    add_a_and_b(b=5, a=3)
except TypeError as e:
    print(e)
````

### 2. F-string Enhancements
Now we can format F strings with = to write name of variable and their values

````python
a = 13
pi = 3.14159265359
f'Value of {a=}, also it takse formating inot account {pi=:.2f} - before it was a={a} or pi={pi:.2f}'
````

### 3. Assignment Expressions (The Walrus Operator)
There is new syntax := that assigns values to variables as part of a larger expression.

````python
a = [1,2,3]
if (n := len(a)) < 5:
    print(f"List is too long ({n} elements, expected <= 10)")
````

````python
import random
def rand_int():
    return random.randint(1,10)
    
[odd_num for x in range(10) if (odd_num := rand_int()) % 2]
````

## Python 3.10
all changes: https://docs.python.org/3/whatsnew/3.10.html

### 1.Structural Pattern Matching
Structural pattern matching has been added in the form of a match statement and case statements of patterns with associated actions. Patterns consist of sequences, mappings, primitive data types as well as class instances. Pattern matching enables programs to extract information from complex data types, branch on the structure of data, and apply specific actions based on different forms of data.

````python
language = 'python'
match language:
    case 'python':
        print('Why does Python live on land?')
        print('Because it is above C level')
    case 'java':
        print('Why did the Java developer quit his job?')
        print('Because he didn’t get arrays.')
    case '.net':
        print('Why dont .NET programmers wear glasses?')
        print('Because they C#.')
    case _:
        print('How many developers does it take to screw in a lightbulb?')
        print('None. It’s a hardware problem.')
````

## Python Enhancement Proposals (PEPs)
It is a document that outlines a proposed new feature, improvement, or change to the Python programming language.

The PEP process is the primary mechanism for proposing major changes to Python, and it provides a framework for the community to discuss and evaluate proposals. Anyone can submit a PEP, and the Python community reviews and discusses the proposal before a decision is made whether to accept or reject it.

### Some Important PEPs

#### PEP 8
link: https://peps.python.org/pep-0008/

PEP 8: Style Guide for Python Code - This PEP outlines the conventions and guidelines for writing clean and readable Python code, covering topics like naming conventions, indentation, and code structure.

#### PEP 257
link: https://peps.python.org/pep-0257/

PEP 257: Docstring Conventions - This PEP defines conventions for writing docstrings, which are used to document Python functions, classes, and modules. Proper docstrings help improve code documentation and make it easier for developers to understand and use Python libraries.

#### PEP 333
link: https://peps.python.org/pep-0333/

PEP 333: Python Web Server Gateway Interface (WSGI) - WSGI is a standard interface between web servers and Python web applications or frameworks. This PEP defines the WSGI specification, enabling interoperability between various web servers and Python web applications.

#### PEP 484
link: https://peps.python.org/pep-0484/

PEP 484: Type Hints - Python's type hinting system allows developers to annotate their code with type information, aiding in static analysis and improving code clarity. PEP 484 introduced type hinting, and subsequent PEPs have refined and expanded upon this feature.

#### PEP 572
link: https://peps.python.org/pep-0572/

PEP 572: Assignment Expressions - This PEP introduced the "walrus operator" (:=), allowing assignments to be made within expressions. It improves code readability by enabling variable assignments within conditional statements and expressions.
